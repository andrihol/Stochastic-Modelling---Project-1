```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

#### a)

```python
# Exact probability P(X(59) > 100) 
import math
t = 59
lam = 1.5 * t
claims = 100

sum = np.exp(-lam)
for i in range(1,claims+1):
    sum += np.exp(-lam)*(lam**i) / math.factorial(i)
probability_exact = 1-sum
print(probability_exact)
```

```python
# Simulate 1000 realizations of X(59)
t = 59
lam = 1.5 * t
sim = 1000

X = np.random.poisson(lam, sim)

plt.hist(X, bins=20, density=True)
plt.xlabel("Number of claims at day 59")
plt.ylabel("Probability density")
plt.title("Simulated distribution of X(59)")
plt.grid()
plt.show()

probability_a = np.mean(X > 100)

print("Estimated probability:", probability_a)
print("Exact probability:", probability_exact)
```

```python
#Simulation of X(t) 0<t<59

for i in range(10):
    claims = np.random.poisson(1.5,59)
    X_i = [0]
    for claim in claims:
        X_i.append(X_i[-1] + claim)
    plt.step(range(60),X_i)

plt.grid()
plt.xlabel("Days")
plt.ylabel("Number of claims")
plt.title("Realizations of X(t) for t in [0,59]")
plt.show()


```

```python
# Simulate the total claim amount Z(59)
gamma = 10
sim = 1000
X_b = np.random.poisson(1.5 * 59, sim)

total_claims = []
for x in X_b:
    claims = np.random.exponential(1 / gamma, x)
    Z = np.sum(claims)
    total_claims.append(Z)

probability_b = np.mean(np.array(total_claims) > 8)

print("Estimated probability:", probability_b)

```

```python
#Simulation Z(t) 0<t<59

for i in range(10):
    total_claims = []
    for x in (np.random.poisson(1.5,59)):
        claims = np.random.default_rng().exponential(1/gamma, x)
        Z = np.sum(claims)
        total_claims.append(Z)

    Z_i = [0]
    for claim in total_claims:
        Z_i.append(Z_i[-1] + claim)
    plt.step(range(60),Z_i)

plt.xlabel("Days")
plt.ylabel("Claim amount (mill. kr)")
plt.title("Realizations of Z(t) for t in [0,59]")
plt.grid()
plt.show()


```
